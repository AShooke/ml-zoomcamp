## Q1. Pandas version

```python
import pandas as pd
import numpy as np

print("Q1. Pandas version:", pd.__version__)
```

## Q2. Data ingestion and record count

```python
url = 'https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv'
df = pd.read_csv(url)

print("Q2. Number of records:", len(df))
df.head()
```

## Q3. Unique fuel types

```python
fuel_types_count = df['fuel_type'].nunique()

print("Q3. Number of unique fuel types:", fuel_types_count)
print("Unique fuel types:", sorted(df['fuel_type'].dropna().unique()))
```

## Q4. Columns with missing values

```python
missing_per_col = df.isnull().sum()
cols_with_missing = (missing_per_col > 0).sum()

print("Q4. Number of columns with missing values:", cols_with_missing)
print("Columns with missing values:")
print(missing_per_col[missing_per_col > 0])
```

## Q5. Maximum fuel efficiency for Asia cars

```python
asia_cars = df[df['origin'] == 'Asia']
fuel_efficiency_column = (
    'fuel_efficiency'
    if 'fuel_efficiency' in asia_cars.columns
    else 'fuel_efficiency_mpg'
)
max_fuel_efficiency = asia_cars[fuel_efficiency_column].max()

print("Q5. Maximum fuel efficiency for Asia cars:", max_fuel_efficiency)
```

## Q6. Horsepower median after mode imputation

```python
horsepower_median_before = df['horsepower'].median()
horsepower_mode = df['horsepower'].mode().iloc[0]
df['horsepower'] = df['horsepower'].fillna(horsepower_mode)
horsepower_median_after = df['horsepower'].median()

if horsepower_median_after > horsepower_median_before:
    median_change = "increased"
elif horsepower_median_after < horsepower_median_before:
    median_change = "decreased"
else:
    median_change = "stayed the same"

print("Q6. Horsepower median before filling:", horsepower_median_before)
print("Q6. Horsepower mode:", horsepower_mode)
print("Q6. Horsepower median after filling:", horsepower_median_after)
print("Q6. The median", median_change + ".")
```

## Q7. Linear regression weights for Asia cars

```python
asia_features = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']]
X = asia_features.iloc[:7].to_numpy()
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

XTX = X.T @ X
XTX_inverse = np.linalg.inv(XTX)
weights = XTX_inverse @ X.T @ y

print("Q7. Sum of the resulting weights:", weights.sum())
```
